### **Step 2**: You first gradio interface

Gradio enables to create front end component easily and declaratively

```python
import gradio as gr

def greet(name):
    return f"Hello, {name}!"

with gr.Blocks() as demo:
    gr.Markdown("# Minimalist Gradio Tutorial")
    name = gr.Textbox(label="Enter your name")
    output = gr.Textbox(label="Output")
    greet_btn = gr.Button("Greet")
    
    greet_btn.click(fn=greet, inputs=name, outputs=output)
    name.change(fn=greet, inputs=name, outputs=output)

demo.launch()
```

Exercise 2 : Navigating the documentation of gradio, create a simple interface to:
- Enable a user to enter a textual data (check `gr.Textbox`)
- Have a button with the string "Compute Sentiment"
- On click, use the `TextBlob` library to compute the sentiment of the text and display it (check `gr.Textbox`)

```python
from textblob import TextBlob

def compute_sentiment(text):
    try:
        blob = TextBlob(text)
        return blob.sentiment.polarity
    except:
        print('Failed to compute')
        return 0


with gr.Blocks() as demo:
    text_to_analyse = gr.Textbox(label="Enter your text")
    sentiment = gr.Number(value = 0)
    
    text_to_analyse.change(compute_sentiment, [text_to_analyse], [sentiment])
    
demo.launch()
```
