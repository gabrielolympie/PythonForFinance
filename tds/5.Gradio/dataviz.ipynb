### **Step 2**: You first gradio interface

Gradio enables to create front end component easily and declaratively

```python
import gradio as gr

def greet(name):
    return f"Hello, {name}!"

with gr.Blocks() as demo:
    gr.Markdown("# Minimalist Gradio Tutorial")
    name = gr.Textbox(label="Enter your name")
    output = gr.Textbox(label="Output")
    greet_btn = gr.Button("Greet")
    greet_btn.click(fn=greet, inputs=name, outputs=output)

demo.launch()
```
