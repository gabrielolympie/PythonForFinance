#### **Step 2: Scrape an Article**
We will target a single news article from Yahoo Finance. The process is simple: you provide the URL, and the library does the rest.

*   Define the URL of the article you want to scrape.
*   Create an `Article` object with this URL.
*   Call the `.download()` method to fetch the page's HTML.
*   Call the `.parse()` method to let `Newspaper3k` process the HTML and identify the key components.
*   Access the extracted data through attributes like `article.title` and `article.text`.

Example url : 'https://finance.yahoo.com/news/pepsis-chips-empire-is-losing-its-shine-144704245.html'

```python
from newspaper import Article
url = "https://finance.yahoo.com/news/pepsis-chips-empire-is-losing-its-shine-144704245.html"

article = Article(url)
article.download()
article.parse()
```

```python
article.authors
```

```python
article.publish_date
```

```python
print(article.text)
```

#### **Part B: The Manual Way with `Requests` and `BeautifulSoup`**

While `Newspaper3k` is convenient, sometimes you need more control to extract very specific pieces of data that it might miss. This requires a two-step process:
1.  **`requests`**: A library that acts like a web browser, downloading the raw HTML source code of a page.
2.  **`BeautifulSoup`**: A library that parses the raw HTML, turning it into a structured object that you can easily search.

#### **Step 4: Parse and Extract Data with `BeautifulSoup`**
Now we have the HTML, but it's a messy string. `BeautifulSoup` will help us navigate it. To know *what* to look for, we must use our browser's developer tools.

**Mini-Task: Find the HTML Tags**
1.  Open the article URL in your browser (Chrome, Firefox, etc.).
2.  Right-click on the article's **title** and select **"Inspect"**.
3.  A developer panel will open, showing you the HTML code. Notice the tag surrounding the title. It's likely an `<h1>` tag with a specific `class` attribute.
4.  Do the same for the main article text. You'll see it's made up of many `<p>` (paragraph) tags inside a parent container.

Now, let's use that knowledge to extract the data.

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html_result)

p_tags = soup.find_all('p')
```
