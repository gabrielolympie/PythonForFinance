#### **Step 3: Fetching Stock Data**
Using what you learned from the `yfinance` documentation, fetch the historical stock data for a few tickers.

*   Define a list of tickers (e.g., `'MSFT'`, `'AAPL'`, `'GOOG'`, `'AMZN'`).
*   Define a start and end date (e.g., one year of data).
*   Use `yf.download()` to get the data.
*   The result is a pandas DataFrame. Print the first 5 rows using the `.head()` method to inspect it.
*   If the result is not a pandas DataFrame, find a way to convert it as a pandas dataframe

Note: yf.download returns a multiindex dataframe by default. You can flatten it by investigating the methods .reset_index (on rows) and the assignation of .columns (on columns)

```python
import yfinance as yf
tickers = ['MSFT', 'AAPL', 'GOOG']

data = yf.download(tickers)

depivoted = data.stack(level=['Price', 'Ticker']).reset_index()
depivoted.columns = ['Date'] + depivoted.columns[1:].tolist()

result = depivoted.pivot(
    index=['Date', 'Ticker'],
    columns='Price',
    values=0
).reset_index()

result.head()
```

```python
import plotly.express as px

# Assuming your dataframe is named 'df'
fig = px.line(
    result,
    x="Date",
    y="Close",
    color="Ticker",  # This creates separate lines for each ticker
    title="Close Price by Date and Ticker",
    labels={"Close": "Price", "Date": "Date"}
)

# Show the figure
fig.show()
```
